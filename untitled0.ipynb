```python
#bank account details
class BankAccount:

    def __init__(self, name, account_number):
        self.name = name
        self.__account_number = account_number
        self.__bank_balance = 1000

    def display(self):
        print("Name:", self.name)
        print("Account Number:", self.__account_number)
        print("Bank Balance:", self.__bank_balance)

    def deposit(self, amount):
        self.__bank_balance += amount
        print("Money successfully deposited")
        print("Updated balance:", self.__bank_balance)

    def withdraw(self, amount):
        if self.__bank_balance < amount:
            print("Insufficient balance")
        else:
            self.__bank_balance -= amount
            print("Money successfully withdrawn")
            print("Updated balance:", self.__bank_balance)

account = BankAccount("Kishore", 123456789)
account.display()
account.deposit(500)
account.withdraw(300)
account.withdraw(2000)
```

```python
from abc import ABC , abstractmethod

class Vehicle(ABC): # Abstract Class
  @abstractmethod
  def start(self):
    pass

  @abstractmethod
  def stop(self):
    pass

class Car(Vehicle): # Concrete class implements the methods of abstract class
  def start(self):
    print("The Car started")
  def stop(self):
    print("The Car stopped")

ob = Car()
ob.start()
```

```python
# Practise Question

# Design a Ride Booking System

# It should have an abstract class named RideBooking
# Having an abstract method named calculate_fare()
# Create two concrete classes BikeRide and CarRide
# These should implement the abstract class method differently
# use python's abc module
```

```python
from abc import ABC , abstractmethod

class Ridebooking(ABC):
  @abstractmethod
  def calculate_fare(self, run_distance):
   pass

class bikeride(Ridebooking):
  def calculate_fare(self, run_distance):
   print("The calculated fare for the ride is:", run_distance*30)

class carride(Ridebooking):
  def calculate_fare(self, run_distance):
   print("The calculated fare for the ride is:", run_distance*50)

bike = bikeride()
car = carride()

bike.calculate_fare(15)
car.calculate_fare(20)
```

```python
# Practise Question

# ATM machine - Use all the 4 pillars

# Build and ATM Machine
# Encapsulation - private balance
# Inheritance - SavingsAccount
# Polymorphism - diff withdrawl behaviours
# Abstraction - abstract the transaction method

# Create a class named as ATM(ABC)
# create a concrete class named Account(ATM), add balance as the attribute
# add get_balance(), set_balance()
# SavingsAccount should inherit from Account
# Display the transaction details
```

```python
from abc import ABC, abstractmethod

class ATM(ABC):
  @abstractmethod
  def transaction(self,balance):
    pass
class Account(ATM):
  def transaction(self,balance):
    self.__balance = 1000
class savings_account(Account):
  def get_balance(self):
    print("account balance:",self.__balance)
  def set_balance(self,type_of_transaction,amount):
    self.type_of_transaction = type_of_transaction
    self.amount = amount
    if type_of_transaction = "deposit":
      self.__balance += amount
      print("money deposited successfully")
    elif type_of_transaction = "withdraw":
      if amount > self.__balance:
        print("insufficient balance")
      else:
        self.__balance -= amount
        print("money withdrawn successfully")
    else:
      print("error")
```

```python
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side


c = Circle(5)
print("Area of Circle:", c.area())

s = Square(4)
print("Area of Square:", s.area())
```
