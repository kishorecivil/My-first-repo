```python
#Grading system
a = int(input("enter marks: "))
if a>100:
 print("wrong input")
if 100>=a>=90:
 print("grade A+")
if 89>=a>=80:
 print("grade A")
if 79>=a>=70:
 print("grade B+")
if 69>=a>=60:
 print("grade B")
if 59>=a>=50:
 print("grade C+")
if 49>=a>=40:
 print("grade C")
if a>=40:
 print("pass")
else:
 print("fail")
```

```python
#calculator
num1 = int(input("enter first number: "))
oper = input("enter operator(+,-,*,/): ")
num2 = int(input("enter second number: "))
if oper == "+":
 print(num1+num2)
elif oper == "-":
 print(num1-num2)
elif oper == "*":
 print(num1*num2)
elif oper == "/":
 print(num1/num2)
else:
 print("wrong input")
```

```python
#n values addition
n = int(input("enter the value: "))
i = 0

while n > 0:
    i = i + n
    n = n - 1
print(i)
```

```python
# factorial
n = int(input("enter the value: "))
i = 1

while n > 0:
  i = i * n
  n = n - 1
print(i)
```

```python
#vowels in word
word = str(input("enter a word: "))
Count = 0
for letter in word:
  if letter in ("aeiouAEIOU"):
   Count+=1
print("number of vowels:",Count)
```

```python
Num = int(input("enter a number: "))
if Num in (2,3,5,7):
 print ("it is a prime number")
elif Num % 2 == 0:
 print ("it is not a prime number")
elif Num % 5 == 0:
 print ("it is not a prime number")
elif Num % 7 == 0:
 print ("it is not a prime number")
else:
 print ("it is a prime number")
```

```python
class Dog:
  def __init__ (xxx,breed,color,age,price):
    xxx.breed = breed
    xxx.color = color
    xxx.age = age
    xxx.price = price
  def display (xxx):
    print("The Dog is",xxx.breed,"of",xxx.color,"color")
    print("It is in age of",xxx.age,"of price",xxx.price)
Doga = Dog("labrador","black",2,15000)
Dogb = Dog("Pomerian","white",3,10000)

Doga.display()
Dogb.display()
```
