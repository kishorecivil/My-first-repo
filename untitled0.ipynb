```python
#Grading system
a = int(input("enter marks: "))
if a>100:
 print("wrong input")
if 100>=a>=90:
 print("grade A+")
if 89>=a>=80:
 print("grade A")
if 79>=a>=70:
 print("grade B+")
if 69>=a>=60:
 print("grade B")
if 59>=a>=50:
 print("grade C+")
if 49>=a>=40:
 print("grade C")
if a>=40:
 print("pass")
else:
 print("fail")
```

```python
#calculator
num1 = int(input("enter first number: "))
oper = input("enter operator(+,-,*,/): ")
num2 = int(input("enter second number: "))
if oper == "+":
 print(num1+num2)
elif oper == "-":
 print(num1-num2)
elif oper == "*":
 print(num1*num2)
elif oper == "/":
 print(num1/num2)
else:
 print("wrong input")
```

```python
#n values addition
n = int(input("enter the value: "))
i = 0

while n > 0:
    i = i + n
    n = n - 1
print(i)
```

```python
# factorial
n = int(input("enter the value: "))
i = 1

while n > 0:
  i = i * n
  n = n - 1
print(i)
```
